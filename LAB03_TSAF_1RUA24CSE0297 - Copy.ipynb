```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = {
    "Month": ["Jan","Feb","Mar","Apr","May","Jun","Jul","Aug","Sep","Oct"],
    "Actual": [100,120,130,150,160,155,170,180,175,190],
    "Forecast": [95,125,128,145,166,150,165,185,180,184]
}
```

```python
df = pd.DataFrame(data)
print(df)
```

```python
df["Error"] = df["Actual"] - df["Forecast"]
```

```python
df["Absolute_Error"] = np.abs(df["Error"])
```

```python
df["Absolute_Percentage_Error"] = (
    df["Absolute_Error"] / df["Actual"]
) * 100
```

```python
print("\nError calculations:")
print(df)
```

```python
mae = df["Absolute_Error"].mean()
```

```python
print("MAE :", round(mae, 2))
```

```python
df["Error"] = df["Actual"] - df["Forecast"]
```

```python
df["Absolute_Error"] = np.abs(df["Error"])
```

```python
df["Squared_Error"] = df["Error"] ** 2
```

```python
df["Absolute_Percentage_Error"] = (
    df["Absolute_Error"] / df["Actual"]
) * 100
```

```python
rmse = np.sqrt(df["Squared_Error"].mean())
print("RMSE:", round(rmse, 2))
```

```python
mape = df["Absolute_Percentage_Error"].mean()

print("MAPE:", round(mape, 2), "%")
```

```python
plt.figure(figsize=(10,5))

plt.plot(df["Month"], df["Actual"], marker="o", label="Actual")

plt.plot(df["Month"], df["Forecast"], marker="o", label="Forecast")

plt.title("Actual vs Forecasted Sales")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.legend()
plt.grid(True)

plt.show()
```

```python
Q1. What is forecast error?
Ans:
Forecast error is the difference between the actual value and the forecasted value.
Error= Actual − Forecast
```

```python
Q2. Why is absolute error used in MAE?
Ans:
Absolute error is used in MAE because forecast errors can be positive or negative. If we directly average the errors, positive and negative errors may cancel each other.
Taking the absolute value makes all errors positive.
Absolute Error= Actual − Forecast
```

```python
Q3. Why does RMSE penalize large errors more than MAE?
Ans:
RMSE squares each error before calculating the average. Therefore,
large errors become much larger after squaring.
```

```python
Q4. What does MAPE tell us?
Ans:
MAPE stands for Mean Absolute Percentage Error.
It tells us the average forecasting error as a percentage of the actual values.
```

```python
Q5. If MAE is 8 sales units, what does it mean?
Ans:
If MAE is 8 sales units, it means that the forecast is wrong by an average of 8 sales units compared with the actual sales.
```

```python
Q6. Can two models have the same MAE but different RMSE? Explain.

Answer:
Yes, two models can have the same MAE but different RMSE.

This is because MAE treats all errors proportionally, whereas RMSE gives more importance to large errors because the errors are squared.
```

```python
Q7. For MAE, RMSE, and MAPE, is a higher or lower value better? Explain.
Ans:
For MAE, RMSE, and MAPE, a lower value is better.Lower values are better because they indicate that the forecast is closer to the actual values.
```

```python
#practice
Actual = [200,220,210,240,250,270]
Forecast = [195,225,215,230,260,260]

Calculate MAE, RMSE, and MAPE using Python without a ready-made accuracy-metric function.
```

```python
import numpy as np
import pandas as pd

actual = [200,220,210,240,250,270]
forecast = [195,225,215,230,260,260]

df = pd.DataFrame({
    "Actual": actual,
    "Forecast": forecast
})

# Calculate error
df["Error"] = df["Actual"] - df["Forecast"]

# Calculate absolute error
df["Absolute_Error"] = np.abs(df["Error"])

# Calculate squared error
df["Squared_Error"] = df["Error"] ** 2

# Calculate absolute percentage error
df["Absolute_Percentage_Error"] = (
    df["Absolute_Error"] / df["Actual"]
) * 100

print(df)

# Calculate MAE
mae = df["Absolute_Error"].mean()

# Calculate RMSE
rmse = np.sqrt(df["Squared_Error"].mean())

# Calculate MAPE
mape = df["Absolute_Percentage_Error"].mean()

print("\nAccuracy Metrics")
print("MAE :", round(mae, 2))
print("RMSE:", round(rmse, 2))
print("MAPE:", round(mape, 2), "%")
```
